```python
import os

if not os.path.exists("lab_data"):
    os.mkdir("lab_data")

os.mkdir("lab_data/input")
os.mkdir("lab_data/output")
os.mkdir("lab_data/logs")
```

```python
import os

for item in os.listdir("lab_data"):
    path = os.path.join("lab_data", item)
    if os.path.isfile(path):
        print(item, "is a file")
    else:
        print(item, "is a directory")
```

```python
import os

os.rename("lab_data/logs", "lab_data/history")
```

```python
with open("lab_data/input/notes.txt", "w") as file:
    file.write("Line 1\n")
    file.write("Line 2\n")
    file.write("Line 3\n")
    file.write("Line 4\n")
    file.write("Line 5\n")
```

```python
with open("lab_data/input/notes.txt", "r") as file:
    lines = file.readlines()

print("Total number of lines:", len(lines))
```

```python
with open("lab_data/output/summary.txt", "w") as file:
    file.writelines(lines)
```
